```python
from os import listdir, path
import hashlib
```

```python
def check_paper_files_consistency(sources, vaccine_type):
    source_titles = [source["title"] for source in sources]
    source_paper_ids = [source["paper_id"] for source in sources]
    all_paper_files = [f.replace(".pdf", "") for f in sorted(listdir(path.join("..", "indicators", vaccine_type, "PDFs")))]
    all_paper_ids = sorted(source_paper_ids)

    missing_in_files = set(all_paper_ids) - set(all_paper_files)
    missing_in_ids = set(all_paper_files) - set(all_paper_ids)

    print("Checking consistency between paper IDs and files...\n")

    print(f"Paper IDs missing from files: {missing_in_files}")
    print(f"Files missing from paper IDs: {missing_in_ids}")
    print(f"Contents match: {set(all_paper_ids) == set(all_paper_files)}")
    print()



def validate_title_md5(sources):
    all_valid = True
    print("Validating title MD5 hashes...\n")
    for source in sources:
        title = source["title"]
        expected_md5 = source["paper_id"]
        
        title_md5 = hashlib.md5(title.encode('utf-8')).hexdigest()

        if title_md5 == expected_md5:
            print(f"✓ Title MD5 matches: {title_md5}")
        else:
            print(f"✗ Title MD5 does not match! Computed: {title_md5}, Expected: {expected_md5}")
            print(f"  Title: {title}")
            all_valid = False
        print()
    
    return all_valid
```
